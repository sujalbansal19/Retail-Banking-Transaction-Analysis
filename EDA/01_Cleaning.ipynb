## Phase 2 : Data Cleaning

### We will clean each DataFrame one by one
#### 1. Lets start with "Account Transactions - df1"

```python
# 01_Missing Values
df1.isna().sum()
df1.count()

"""The linked_transaction_id column contains missing values because it is used only for loan disbursement and repayment transactions to link the
loan account with the savings account transaction. These missing values are valid and business-driven, so no imputation was performed, and the 
column was retained."""
```

```python
# 02_Duplicate Values
df1.duplicated().sum()  #No duplicates found
```

```python
# 03_Checking Data Types
df1.dtypes

df1["created_time"]=pd.to_datetime(df1["created_time"])
# pd.to_datetime(df["created_time"], errors="coerce") if error is there that becomes 'NaT'
```

```python
# 04_Data Validation - Checking col-wise whether data is valid

# id's col already chked if null or not

# linked_transaction_id col - type handling it from NaN-> pd.Na
df1["linked_transaction_id"]=df1["linked_transaction_id"].fillna(value=pd.NA)

# created_time - to be in btw year 2020-2025
df1["year"] = df1["created_time"].dt.year
print("Count of Transaction out of 2020-2025 range : ",(~df1["year"].between(2020,2025)).sum())
# print(((df1["year"]<2020)|(df1["year"]>2025)).sum())


# transaction_type col
print("\nChecking categories",df1["transaction_type"].value_counts()) # values dekhke rechk kr skte, koi lettering mistakes hai, (eg:savings SAVINGS)

# transaction_code - comparing it with transaction table 
c=df4["transaction_code"].unique()
s=~df1["transaction_code"].isin(c)
print("\nCount of Transaction codes not there in df4: ",s.sum())
 
# channel - comparing it will transaction table 
x=df4["channel"].str.split(",") # spliting the values separated by comma
x=x.explode().str.strip().unique()  
# explode seperates each row into individual, i.e., atm branch, becomes 2 seperate rows
# strip remove unwanted spaces from front nd back
# we didnt edit the df, created a new series

y=~df1["channel"].isin(x)
print("\nCount of channels not there in df4 : ",y.sum())


# amount col
print("\nValues less than zero : ",(df1["amount"]<0).sum())  # Here values less than zero become true, i.e., '1', and sum returns count of true
```

```python
# THIS WAS THE 2ND DIFFICULTY FACED
# 05_Checking Outliers

#First visual check 
df1.select_dtypes(include="number").columns # list of all numerical cols present

sns.boxplot(x=df1["amount"])
plt.show()


# Checking Skewness
print("\n Skew Value : ",df1["amount"].skew())  

"""The transaction amount distribution was found to be highly positively skewed, with a skewness value of 58.60,i.e., most transactions are 
relatively small, while a smaller number of transactions are extremely large. The boxplot also showed a long upper tail, indicating the presence 
of numerous high-value transactions. Since these values may represent legitimate banking transactions,they were not removed 
solely based on statistical outlier detection.
Note: Highly right skewed so didnt use IQR method here."""




# Percentile Analysis
pd.set_option("display.float_format", "{:.2f}".format) # changing the scientific display format to float

df1["amount"].describe(percentiles=[0.50,0.75,0.90,0.99,0.999])

print("\nPercentile Analysis : ")
print(df1.groupby("transaction_code")["amount"].describe(percentiles=[0.90, 0.95, 0.99, 0.999]))
"""Percentile analysis was used to identify extreme-value transactions. These transactions were further examined against transaction codes, 
transaction types, channels, and business rules to distinguish legitimate high-value transactions from potential data-quality issues."""
```

#### 2. Now, "Bank Accounts - df2"

```python
# 01_Checking null values
print("\nCount of Null values : \n",df2.isna().sum())

"""The missing values in loan_amount, term_months, and interest_rate are business-driven. These fields are not applicable to all account types, 
so the null values are genuine and no imputation was performed.And, loan_amount is applicable specifically to Loan accounts, 
while term_months and interest_rate are applicable to Loan and Fixed Deposit accounts."""

df2["loan_amount"]=df2["loan_amount"].fillna(0)
df2["term_months"]=df2["term_months"].fillna(0)
df2["interest_rate"]=df2["interest_rate"].fillna(0)

print("\nCount of Null values after cleaning : \n",df2.isna().sum())


# 02_Checking for duplicates
print("\nCount of duplicates : ",df2.duplicated().sum())


# 03_Checking Data Types 
df2["creation_date"]=pd.to_datetime(df2["creation_date"])
print("\nDatatypes : \n",df2.dtypes)


# 04_Data Validation
df2["account_type"].unique()
df2["account_status"].unique()

# numerical cols
print("\nCount of values less than zero for 'balance' : ",(df2["balance"]<0).sum())
print("Count of values less than zero for 'loan_amount' : ",(df2["loan_amount"]<0).sum())

# checking date cols year btw 2020-2025
x=~df2["creation_date"].dt.year
x=x.between(2020,2025).sum()
print("Checking for year in date column",x)



# 05_Detecting Outliers

# Using visuals
num=df2.select_dtypes(include="number").columns

plt.figure(figsize=(12,8))

for i,col in enumerate(num):
    plt.subplot(2,2,i+1)
    sns.boxplot(x=df2[col])
    plt.title(col)
plt.tight_layout()
plt.show()
# This tells us that only balance column has outliers

# Percentile Analysis
print("Skewness is : ",df2["balance"].skew())
df2["balance"].describe(percentiles=[0.25,0.50,0.75,0.90,0.99,0.999])
```

# Phase 3 : Feature Engineering

```python
# Adding age column to Customer profile table
today = pd.Timestamp.today()
df3["age"] = (today - df3["birth_date"]).dt.days // 365

# Adding age group col to Customer profile table
bins=[17,25,45,60,100]
labels=['Young Adults','Middle Aged','Older Adults','Seniors']
df3["age_group"] = pd.cut(df3["age"],bins=bins,labels=labels)
```
